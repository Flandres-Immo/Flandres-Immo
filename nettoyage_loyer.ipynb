```python
import pandas as pd
df_loyer = pd.read_csv("./data/Indicateurs loyers appartement 1 ou 2 pieces.csv", 
                       encoding="cp1252", 
                       sep=";")
df_loyer.head() 
```

```python
df_loyer.shape
```

```python
df_loyer.head(5)
```

```python
df_loyer.duplicated().sum()
```

```python
df_loyer["loypredm2"] = (
    df_loyer["loypredm2"]
    .str.replace(",", ".", regex=False)
    .astype(float)
    .round(2)
)
```

```python
colonnes_float = [
    "lwr.IPm2",
    "upr.IPm2",
    "R2_adj"
]

for colonne in colonnes_float:
    df_loyer[colonne] = (
        df_loyer[colonne]
        .str.replace(",", ".", regex=False)
        .astype(float)
        .round(2)
    )
```

```python
df_loyer.head(5)
```

```python
df_loyer["id_zone"] = pd.to_numeric (df_loyer["id_zone"], errors="coerce").astype("Int64")
```

```python
df_loyer["INSEE_C"] = pd.to_numeric (df_loyer["INSEE_C"], errors="coerce").astype("Int64")
```

```python
df_loyer["EPCI"] = pd.to_numeric (df_loyer["EPCI"], errors="coerce").astype("Int64")
```

```python
df_loyer["DEP"] = pd.to_numeric (df_loyer["DEP"], errors="coerce").astype("Int64")
```

```python
df_loyer.dtypes
```

```python
print("NaN :", df_loyer.isna().sum().sum())

print("null :", df_loyer.astype("string").apply(
    lambda col: col.str.strip().str.lower().eq("null").sum()
).sum())

print("Cellules vides 0 :", df_loyer.astype("string").apply(
    lambda col: col.str.strip().eq("0").sum()
).sum())
```
