```python
import os
import sys
import numpy as np
import pandas as pd

pd.set_option('display.max_columns', 50)
pd.set_option('display.max_rows', 20)

# Target
path_source = "../Data/Silver/"

# File Ille et Vilaine
name_iev = "Ille et Vilaine"
filename_iev = "dvf Ille et Vilaine.csv"
# File Loire Atlantique
name_la = "Loire Atlantique"
filename_la  = "dvf Loire Atlantique.csv"
# File Rhone
name_r = "Rhone"
filename_r  = "dvf Rhone.csv"
# Indicateurs loyers maison.csv
name_lm = "Indicateurs loyers maison"
filename_lm = "Indicateurs loyers maison.csv"

lst_names = [name_iev, name_la, name_r, name_lm]
lst_filenames = [filename_iev, filename_la, filename_r, filename_lm]
lst_encoding = ['utf-8'] * 3 + ["cp1252"]
```

```python
dfs = [pd.read_csv(path_source + filename, encoding=lst_encoding[i]) for i, filename in enumerate(lst_filenames)]
df_iev, df_la, df_r, df_lm = dfs

for i, df in enumerate(dfs):
    print(f'SHAPE {lst_names[i]:50} : {df.shape}')

colonnes = ['id_zone', 'INSEE_C', 'EPCI', 'DEP', 'REG',	'nbobs_com', 'nbobs_mail']
df_lm[colonnes] = df_lm[colonnes].astype('Int64')

display(dfs[0].head(5))
display(df_lm.head(5))
df_lm.dtypes
```
