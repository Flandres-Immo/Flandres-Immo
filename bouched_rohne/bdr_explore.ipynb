```python
df = pd.read_csv(
    "../raw_data/dvf Bouche du Rhone.csv",
    dtype={"code_postal": "string"}
)
df.head()
```

```python
# suppression des colonnes indésirables

colonnes_supprimées = [ 'adresse_numero',
                        'adresse_suffixe',
                        'adresse_nom_voie',
                        'adresse_code_voie',
                        'ancien_code_commune',
                        'ancien_nom_commune',
                        'id_parcelle',
                        'ancien_id_parcelle',
                        'numero_volume',
                        'lot1_numero',
                        'lot1_surface_carrez',
                        'lot2_numero',
                        'lot2_surface_carrez',
                        'lot3_numero',
                        'lot3_surface_carrez',
                        'lot4_numero',
                        'lot5_numero',
                        'lot5_surface_carrez',
                        'nature_culture_speciale',
                        'nombre_lots',
                        'code_nature_culture',
                        'nature_culture',
                        'code_nature_culture_speciale',
                        'longitude',
                        'latitude',
                        'section_prefixe',
                        'numero_disposition',
                        'lot4_surface_carrez'
                        ]

df = df.drop(columns=colonnes_supprimées, errors="ignore")
```

```python
# suppression des espaces vides de part et d'autre des termes

for col in df.select_dtypes(include=["object", "string"]).columns:
    df[col] = df[col].map(
        lambda value: value.strip() if isinstance(value, str) else value
    )
```

```python
# Repérer les espaces superflus et caractères invisibles dans les colonnes textuelles
# sans modifier les données.

text_cols = df.select_dtypes(include=["object", "string"]).columns
pattern = r"(?:^\s|\s$|[\t\r\n\u00A0\u2007\u202F\u200B\uFEFF]|\s{2,})"
for col in text_cols:
    values = df[col].astype("string")
    mask = values.str.contains(pattern, regex=True, na=False)

    if mask.any():
        examples = df.loc[mask, col].dropna().drop_duplicates().head(5)
        print(f"\n{col}: {int(mask.sum())} rows")
        print([repr(value) for value in examples])
```

```python
df = df.drop_duplicates()
print(df.duplicated().sum())
```

```python
# suppression des lignes qui contiennent "Dépendance"
# de la colonne type_local

mask = df["type_local"].str.contains("Dépendance", case=False, na=False)
print("Rows to remove:", mask.sum())

df = df.loc[~mask].copy()
```

```python
# suppression des lignes qui contiennent "Local industriel. commercial
# de la colonne type_local

mask = df["type_local"].str.contains("Local industriel. commercial ou assimilé", case=False, na=False, regex=False)
print("Rows to remove:", mask.sum())

df = df.loc[~mask].copy()
```

```python
df.duplicated().sum()
```

```python
# verification de la consistance des colonnes type_local et code_type_local

type_name = df["type_local"].astype("string").str.strip().str.casefold()
code = pd.to_numeric(df["code_type_local"], errors="coerce")

mismatch = (
    (type_name.eq("maison") & code.ne(1.0)) |
    (type_name.eq("appartement") & code.ne(2.0))
)

print("Mismatched rows:", int(mismatch.sum()))
df.loc[mismatch, ["type_local", "code_type_local"]].head(20)
```

```python
# suppression de la colonne code_type_local 

df = df.drop(columns=["code_type_local"])
```

```python
# valeurs manquantes 
valeurs_manquantes = pd.DataFrame({
    "nb_valeurs_manquantes": df.isna().sum(),
    "pct_valeurs_manquantes": (df.isna().mean() * 100).round(2)
})

valeurs_manquantes
```

```python
text_columns = [
    "id_mutation",
    "nature_mutation",
    "code_postal",
    "code_commune",
    "nom_commune",
    "code_departement",
    "type_local",
]

df[text_columns] = (
    df[text_columns]
    .astype("string")
    .fillna("Non renseigné")
)
```

```python
# les valeurs manquantes remplacées par 0
df = df.fillna(0)
```

```python
# vérification 
df.isna().sum().sum()
```

```python
repeat_header = (
    df["id_mutation"].astype("string").str.strip().eq("id_mutation")
)
df = df.loc[~repeat_header].copy()
```

```python
# vérification de la date

original_dates = df["date_mutation"]
parsed_dates = pd.to_datetime(original_dates, errors="coerce")

invalid = original_dates.notna() & parsed_dates.isna()
print("Unparsed non-missing dates:", int(invalid.sum()))

df["date_mutation"] = parsed_dates
```

```python
df["code_postal"] = df["code_postal"].str.replace(r"\.0$", "", regex=True)
```

```python
numeric_cols = [
    "valeur_fonciere",
    "surface_reelle_bati",
    "nombre_pieces_principales",
    "surface_terrain",
]

df[numeric_cols] = df[numeric_cols].apply(pd.to_numeric, errors="raise")
```
